# Validation and preparation of the MDT data of `munich_ablation`

**Purpose.** This notebook validates the cells and the minimization-of-drive-tests (MDT)
records in `lab/data/munich_ablation/raw/`, and writes them to `lab/data/munich_ablation/processed/` in the Apache Parquet format.
The notebook does not clean, filter, or change a value. If one rule fails, the notebook stops
and writes no file.

**Inputs.**

| File | Content |
|---|---|
| `lab/configs/config.yaml` | The configuration. `utils.load_config` composes it with the other files of `lab/configs/`. The notebook reads the limits of the RSRP report, `seed`, and `scenario.split` from it. |
| `lab/data/munich_ablation/raw/cells.csv` | One row for each cell |
| `lab/data/munich_ablation/raw/mdt.csv` | One row for each MDT record, one `rsrp_<cell>` column for each cell |

**Method.** `preprocessing.validate_cells` and `preprocessing.validate_mdt` count the rows that break
each rule. `preprocessing.mdt_to_long` changes the records to the long format.
`preprocessing.split_mdt` divides the records into the train, validation, and test parts
with the method of `scenario.split`. The notebook then writes the Parquet files and reads them again to verify them.

**Outputs.**

| File | Content |
|---|---|
| `lab/data/munich_ablation/processed/train.parquet` | The train records: one row for each received (record, cell) pair |
| `lab/data/munich_ablation/processed/validation.parquet` | The validation records, in the same format |
| `lab/data/munich_ablation/processed/test.parquet` | The test records, in the same format |
| `lab/data/munich_ablation/processed/cells.parquet` | The content of `cells.csv` |

**Requirements.** Run `task scenario -- scenario=munich_ablation` first. The notebook does not need a GPU.


## 1. Setup

```python
import pandas as pd
from omegaconf import OmegaConf

from preprocessing import (
    MDT_POSITION_COLUMNS,
    SPLIT_NAMES,
    mdt_to_long,
    read_cells,
    rsrp_column,
    split_mdt,
    validate_cells,
    validate_mdt,
)
from utils import PROJECT_ROOT, load_config, save_table, scenario_dirs, setup_notebook

cfg = load_config("scenario=munich_ablation")
RAW_DIR = scenario_dirs(cfg.scenario.name).raw
PROCESSED_DIR = scenario_dirs(cfg.scenario.name).processed
device = setup_notebook(cfg)
```

## 2. Input data

`read_cells` reads `cells.csv` and stops if a column is missing. `mdt.csv` is read with the
`timestamp` column as a datetime. Table 1 gives the size of the inputs.

```python
cells = read_cells(RAW_DIR / "cells.csv")
cells_frame = pd.read_csv(RAW_DIR / "cells.csv")
mdt = pd.read_csv(RAW_DIR / "mdt.csv", parse_dates=["timestamp"])
rsrp_columns = [rsrp_column(cell.name) for cell in cells]

save_table(
    pd.Series(
        {
            "Cells": len(cells),
            "Sites": cells_frame["site"].nunique(),
            "MDT records": len(mdt),
            "Columns of mdt.csv": mdt.shape[1],
            "Position columns": len(MDT_POSITION_COLUMNS),
            "RSRP columns": len(rsrp_columns),
            "Received (record, cell) pairs": int(mdt[rsrp_columns].notna().sum().sum()),
        }
    )
    .to_frame("Value")
    .style.set_caption("Table 1. Size of the input data."),
    "preprocessing_inputs",
)
```

## 3. Validation

Table 2 gives the rules. A rule applies to one row of a file. For each rule, the validation
counts the rows that break it. All counts must be zero.

| Rule | File | Source |
|---|---|---|
| The columns are `timestamp`, `x`, `y`, `z`, and one `rsrp_<cell>` column for each cell | `mdt.csv` | Format of `mdt.csv` (`radio_calibration.core.mdt`). A failure stops the validation with a `ValueError`. |
| `timestamp` has a datetime type | `mdt.csv` | Format of `mdt.csv`. A failure stops the validation with a `ValueError`. |
| Each cell name is unique | `cells.csv` | The long format identifies a cell by its name. |
| Each cell position is finite | `cells.csv` | The distance from a cell to a UE must exist. |
| Each record has a timestamp | `mdt.csv` | Format of `mdt.csv` |
| Each position is finite | `mdt.csv` | The distance from a cell to a UE must exist. |
| Each record has a minimum of one cell | `mdt.csv` | Rule of the generator: it samples a record again if no cell remains. |
| No RSRP is below `scenario.measurement.min_rsrp_dbm` | `mdt.csv` | Rule of the generator: it removes a cell below this value. |
| Each RSRP is a multiple of `scenario.measurement.rsrp_resolution_db` | `mdt.csv` | Resolution of the SS-RSRP report (3GPP TS 38.133) |
| No record is a copy of another record | `mdt.csv` | The records are statistically independent. A copy shows an error of the export. |

**Table 2.** Validation rules.

A missing RSRP is not an error. It shows that the UE did not receive the cell, or that the
RSRP was below `scenario.measurement.min_rsrp_dbm`.

```python
results = {
    **validate_cells(cells),
    **validate_mdt(
        mdt,
        cells,
        cfg.scenario.measurement.min_rsrp_dbm,
        cfg.scenario.measurement.rsrp_resolution_db,
    ),
}
for name, count in results.items():
    print(f"{'PASS' if count == 0 else 'FAIL'}  {name}: {count} rows")
assert all(count == 0 for count in results.values()), "The data is not valid. No file is written."
```

## 4. Long format

The calibration compares one measured RSRP with one simulated RSRP. The unit of the data is
thus one (record, cell) pair. Let $r$ be the index of a record in `mdt.csv`, and let
$\mathcal{C}_r$ be the set of cells with a reported RSRP in record $r$. The long format has
one row for each pair $(r, c)$ with $c \in \mathcal{C}_r$:

$$\left(r,\ t_r,\ x_r,\ y_r,\ z_r,\ c,\ \widehat{\mathrm{RSRP}}_{r,c}\right).$$

The number of rows is thus $\sum_r |\mathcal{C}_r|$. A pair with a missing RSRP gives no
row. The product of `record` and `cells.parquet` gives all pairs again. Table 3 gives the
columns, and Table 4 gives the first rows.

```python
long = mdt_to_long(mdt, cells)
assert len(long) == mdt[rsrp_columns].notna().sum().sum()
assert long["record"].nunique() == len(mdt)

descriptions = {
    "record": "Index of the record in mdt.csv, from 0",
    "timestamp": "Start of the time slot of the record",
    "x": "Reported UE position on the x axis of the scene (m)",
    "y": "Reported UE position on the y axis of the scene (m)",
    "z": "UE position on the z axis of the scene (m)",
    "cell": "Name of the cell, as in cells.parquet",
    "rsrp_dbm": "Reported RSRP of the cell (dBm)",
}
display(
    save_table(
        pd.DataFrame(
            {"Type": long.dtypes.astype(str), "Description": pd.Series(descriptions)}
        ).style.set_caption("Table 3. Columns of the long format."),
        "preprocessing_columns",
    )
)
save_table(
    long.head(len(cells))
    .style.set_caption("Table 4. First rows of the long format.")
    .format(precision=1),
    "preprocessing_head",
)
```

## 5. Split

The model learns from one part of the records and measures its error on a different part.
`split_mdt` uses the method `scenario.split.method` with the seed `seed`:

- `random`: the records go to the parts at random. The train part gets the first
  $\lfloor f_\mathrm{train} N \rfloor$ records of a random order, the validation part the
  subsequent $\lfloor f_\mathrm{val} N \rfloor$ records, and the test part all other records,
  with the fractions `scenario.split.fractions`. A test record can thus be near a train record.
- `site`: the test part gets all (record, cell) pairs of the cells of `scenario.split.test_sites`.
  The other pairs go to the train and validation parts by record, with
  `scenario.split.validation_fraction`. No train pair is for a test cell.
- `spatial`: the records go to the parts by square blocks of `scenario.split.block_size_m` of the
  reported position, with the fractions `scenario.split.fractions` of the blocks. No test record
  has a train record in its block.

All (record, cell) pairs of one record go to the same part, except for the method `site`. The
`record` values do not change, thus a row of each part still points to its row of `mdt.csv`.
Table 5 gives the size of the parts.

```python
split = OmegaConf.to_container(cfg.scenario.split, resolve=True)
parts = split_mdt(long, cells, split, seed=cfg.seed)
assert list(parts) == list(SPLIT_NAMES)
assert sum(len(part) for part in parts.values()) == len(long)

print(f"Split method: {split['method']}")
save_table(
    pd.DataFrame(
        {
            "Records": [part["record"].nunique() for part in parts.values()],
            "(record, cell) pairs": [len(part) for part in parts.values()],
            "Cells": [part["cell"].nunique() for part in parts.values()],
        },
        index=list(parts),
    ).style.set_caption("Table 5. Size of the parts."),
    "preprocessing_parts",
)
```

## 6. Output

The notebook writes one file for each part and one file for the cells. It reads each file
again and compares it with the frame in memory. Table 6 gives the files.

```python
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
outputs = {f"{name}.parquet": part for name, part in parts.items()}
outputs["cells.parquet"] = cells_frame
rows = []
for name, frame in outputs.items():
    path = PROCESSED_DIR / name
    frame.to_parquet(path, index=False)
    pd.testing.assert_frame_equal(pd.read_parquet(path), frame)
    rows.append(
        {
            "File": path.relative_to(PROJECT_ROOT).as_posix(),
            "Rows": len(frame),
            "Columns": frame.shape[1],
            "Size (kB)": path.stat().st_size / 1e3,
        }
    )
save_table(
    pd.DataFrame(rows)
    .style.set_caption("Table 6. Files in lab/data/munich_ablation/processed.")
    .format({"Size (kB)": "{:.1f}"})
    .hide(axis="index"),
    "preprocessing_files",
)
```

All rules of Table 2 have a count of zero, else the notebook stops in Section 3. The

into three parts. The notebooks 03, 04, and 05 of `munich_ablation` read the files of `lab/data/munich_ablation/processed`.
